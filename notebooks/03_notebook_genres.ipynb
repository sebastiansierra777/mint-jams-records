```python
import sqlite3
import pandas as pd
from pathlib import Path
```

```python
project_root = Path.cwd().parent
db_path = project_root / "data" / "mint_jams.db"
conn = sqlite3.connect(db_path)
```

**QUERY: TOTAL SPENT BY ALBUM BY GENRE**

```python
query = """
    SELECT
        albums.title,
        albums.genre,
        SUM(sale_items.quantity * sale_items.unit_price) AS total_spent
FROM albums
JOIN sale_items ON albums.album_id = sale_items.album_id
GROUP BY
    albums.album_id,
    albums.title,
    albums.genre
ORDER BY
    total_spent DESC;
"""

df_top_genres = pd.read_sql_query(query, conn)
display(df_top_genres)
```

**QUERY: TOTAL SPENT BY GENRE**

```python
query = ("""
    SELECT
        albums.genre,
        SUM(sale_items.quantity * sale_items.unit_price) AS total_spent
    FROM albums
    JOIN sale_items ON albums.album_id = sale_items.album_id
    GROUP BY
        albums.genre
    ORDER BY
        total_spent DESC;
""")

df_top_genres_2 = pd.read_sql_query(query, conn)
display(df_top_genres_2)
```

**PLOT: SIMPLE PANDAS HTML / CSS PLOT**

```python
(df_top_genres_2
    .set_index('genre')
    .style
    .bar(subset=['total_spent'], color='#0078D7', vmin=0)
    .format("${:.2f}")
)
```
