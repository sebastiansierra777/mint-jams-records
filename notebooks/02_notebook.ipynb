```python
import sqlite3
import pandas as pd
from pathlib import Path
```

**DATABASE CONNECTION**

```python
project_root = Path.cwd().parent
db_path = project_root / "data" / "mint_jams.db"
conn = sqlite3.connect(db_path)
```

**QUERY: SHOW ALL SALES ORGANIZED BY CUSTOMER**

```python
pd.read_sql_query("""
    SELECT
        customers.first_name AS customer_name,
        customers.last_name AS customer_lastname,
        sales.sale_id,
        sales.sale_date,
        sales.payment_method
    FROM customers
    JOIN sales ON customers.customer_id = sales.customer_id;
""", conn)
```

**QUERY: TOTAL SPENT BY CUSOMTER, RANKED**

```python
pd.read_sql_query("""
    SELECT
        customers.first_name,
        customers.last_name,
        SUM(sale_items.quantity * sale_items.unit_price) AS total_spent
    FROM customers
    JOIN sales
        ON customers.customer_id = sales.customer_id
    JOIN sale_items
        ON sales.sale_id = sale_items.sale_id
    GROUP BY
        customers.first_name,
        customers.last_name
    ORDER BY
        total_spent DESC;
""", conn)
```

**QUERY: TOTAL SPENT BY ALBUM, RANKED**

```python
pd.read_sql_query("""
    SELECT
        albums.title,
        artists.name AS artist_name,
        albums.unit_price,
        SUM(sale_items.quantity * sale_items.unit_price) AS total_spent
    FROM albums
    JOIN sale_items ON albums.album_id = sale_items.album_id
    JOIN artists ON albums.artist_id = artists.artist_id
    GROUP BY
        albums.title
    ORDER BY
        total_spent DESC;
""", conn)
```
